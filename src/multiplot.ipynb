```python
# matplotlib wrappers :)

def multiplot(data, x_col, y_col, colors=['royalblue', 'mediumseagreen', 'orange', 'firebrick', 'gold', 'mediumpurple']):
    for i in range(len(data)):
        df = data[i]
        plt.plot(df[x_col], df[y_col], color=colors[i], label = 'trial ' + str(i + 1))
    plt.xlabel(x_col)
    plt.ylabel(y_col)
    plt.title(y_col + ' vs. ' + x_col)
    plt.legend()
    plt.show()


def avgplot(data, x_col, y_col, colors=[]):
    all_col = []
    for i in range(len(data)):
        all_col.append(data[i][y_col])
    all_col = np.array(all_col)
    all_col = np.mean(all_col, axis=0)
    for i in range(len(data)):
        df = data[i]
        plt.plot(df[x_col], df[y_col], color=colors[0])
    plt.plot(data[0][x_col], all_col, color=colors[1], label='average')
    plt.xlabel(x_col)
    plt.ylabel(y_col)
    plt.title(y_col + ' vs. ' + x_col)
    plt.legend()
    plt.show()
```

```python
data = []

for file in os.listdir(DPATH):
    load = pd.read_csv(os.path.join(DPATH, file), header=0)
    data.append(load)
```

```python
multiplot(data, 'time', 'ldepth')
```

```python
avgplot(data, 'Time (s)', 'Left Cavity Depth (mm)', colors=['lightsteelblue', 'navy'])
```
